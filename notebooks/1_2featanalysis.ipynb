```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

FEATURE_CSV = "../data/metadata/speech_features.csv"

df = pd.read_csv(FEATURE_CSV)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
print("Number of samples:", len(df))
print("Number of features:", len(df.columns))

print("\nMissing values:")
missing = df.isnull().sum()
print(missing[missing > 0])

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("\nDataset distribution:")
print(df["dataset"].value_counts())

print("\nEmotion distribution:")
print(df["emotion"].value_counts())

print("\nSpeaker distribution:")
print(df["speaker"].nunique())
```

```python
feature_cols = [
    "pitch_mean",
    "pitch_median",
    "pitch_min",
    "pitch_max",
    "pitch_std",
    "pitch_range",
    "energy_mean",
    "energy_std",
    "energy_min",
    "energy_max",
    "energy_range",
    "zcr_mean",
    "zcr_std",
    "zcr_min",
    "zcr_max",
    "jitter",
    "shimmer"
]

print(df[feature_cols].describe().T)
```

```python
print(df[feature_cols].isnull().sum())
```

```python
missing_js = df[
    df["jitter"].isna() |
    df["shimmer"].isna()
]

print(missing_js[
    ["file", "dataset", "speaker", "emotion", "jitter", "shimmer"]
])
```

```python
print("Missing jitter:", df["jitter"].isna().sum())
print("Missing shimmer:", df["shimmer"].isna().sum())
```

```python
print(
    df[
        ["pitch_mean", "pitch_min", "pitch_max",
         "energy_mean", "zcr_mean",
         "jitter", "shimmer"]
    ].describe().T
)
```

```python
print("\nPitch < 50 Hz:")
print((df["pitch_mean"] < 50).sum())

print("\nPitch > 300 Hz:")
print((df["pitch_mean"] > 300).sum())

print("\nInvalid pitch ranges:")
print((df["pitch_range"] < 0).sum())
```

```python
import pandas as pd
from pathlib import Path

DEAM_PATH = Path("../data/music/DEAM")

valence_path = (
    DEAM_PATH /
    "DEAM_Annotations" /
    "annotations" /
    "annotations per each rater" /
    "dynamic (per second annotations)" /
    "valence" /
    "18.csv"
)

arousal_path = (
    DEAM_PATH /
    "DEAM_Annotations" /
    "annotations" /
    "annotations per each rater" /
    "dynamic (per second annotations)" /
    "arousal" /
    "18.csv"
)

valence = pd.read_csv(valence_path)
arousal = pd.read_csv(arousal_path)

print("VALENCE")
print(valence.head())
print(valence.columns)

print("\nAROUSAL")
print(arousal.head())
print(arousal.columns)
```

```python
print("Valence shape:", valence.shape)
print("Arousal shape:", arousal.shape)

print("\nValence row means:")
print(valence.mean(axis=1))

print("\nArousal row means:")
print(arousal.mean(axis=1))
print("\nValence overall mean:", valence.values.mean())
print("Arousal overall mean:", arousal.values.mean())
```

```python
valence_consensus = valence.mean(axis=0)
arousal_consensus = arousal.mean(axis=0)
```
